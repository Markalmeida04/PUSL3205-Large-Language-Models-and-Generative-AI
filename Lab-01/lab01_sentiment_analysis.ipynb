```python
import pandas as pd
import numpy as np
import re
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load and combine the datasets
files = ['amazon_cells_labelled.txt', 'imdb_labelled.txt', 'yelp_labelled.txt']
dfs = [pd.read_csv(f, sep='\t', names=['sentence', 'label']) for f in files]
df = pd.concat(dfs, ignore_index=True)

# 2. Basic text preprocessing
def clean_text(text):
    text = text.lower()                          # Convert to lowercase
    text = re.sub(r'[^a-z0-9\s]', '', text)      # Remove punctuation/special chars
    text = re.sub(r'\s+', ' ', text).strip()     # Remove extra whitespace
    return text

df['clean_text'] = df['sentence'].apply(clean_text)

# Compare original vs preprocessed
print(df[['sentence', 'clean_text', 'label']].head(10))

# 3. Explore the text data
print("\nPositive Examples:\n", df[df['label'] == 1]['clean_text'].head(3).tolist())
print("\nNegative Examples:\n", df[df['label'] == 0]['clean_text'].head(3).tolist())

# Frequently occurring words
pos_words = Counter(" ".join(df[df['label'] == 1]['clean_text']).split()).most_common(10)
neg_words = Counter(" ".join(df[df['label'] == 0]['clean_text']).split()).most_common(10)
print("\nTop Positive Words:", pos_words)
print("Top Negative Words:", neg_words)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# 4. Split the dataset (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'], df['label'], test_size=0.2, random_state=42
)

# 5. Create BOW representation
vectorizer = CountVectorizer(stop_words='english')
X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)

# 6. Train a ML model (Logistic Regression)
model = LogisticRegression()
model.fit(X_train_bow, y_train)
print("Model training complete.")
```
